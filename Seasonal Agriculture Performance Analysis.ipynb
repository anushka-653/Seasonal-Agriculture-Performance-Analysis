## Q1. How does agricultural production vary across seasons?

```python
season_production = df.groupby('Season')['Production_Tonnes'].mean().sort_values(ascending=False)
print(season_production)
```

```python
season_production.plot(kind='bar', figsize=(8,5))
plt.title('Average Agricultural Production by Season')
plt.xlabel('Season')
plt.ylabel('Average Production (Tonnes)')
plt.xticks(rotation=0)
plt.show()
```

## Q2. How does average crop yield vary across seasons?

```python
season_yield = df.groupby('Season')['Yield_Tonnes_Ha'].mean().sort_values(ascending=False)
print(season_yield)
```

```python
season_yield.plot(kind='bar', figsize=(8,5))
plt.title('Average Crop Yield by Season')
plt.xlabel('Season')
plt.ylabel('Yield (Tonnes/Ha)')
plt.xticks(rotation=0)
plt.show()
```

## Q12. Which crops perform best in each season?

```python
crop_season_yield = df.pivot_table(
    values='Yield_Tonnes_Ha',
    index='Crop',
    columns='Season',
    aggfunc='mean'
)
print(crop_season_yield)
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(crop_season_yield, annot=True,fmt='.2f')
plt.title('Average Crop Yield Across Seasons')
plt.xlabel('Season')
plt.ylabel('Crop')
plt.show()
```

```python
crop_season_profit = df.pivot_table(
    values='Profit_INR',
    index='Crop',
    columns='Season',
    aggfunc='mean'
)
print(crop_season_profit)
```

## Q13. Do seasonal patterns differ across states?

```python
state_season_yield = df.pivot_table(
    values='Yield_Tonnes_Ha',
    index='State',
    columns='Season',
    aggfunc='mean'
)
print(state_season_yield)
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(
    state_season_yield,
    annot=True,
    fmt='.2f'
)
plt.title('Average Yield by State and Season')
plt.xlabel('Season')
plt.ylabel('State')
plt.show()
```

## Q14. Overall Seasonal Performance Comparison

```python
overall_season = df.groupby('Season')[
    [
        'Yield_Tonnes_Ha',
        'Production_Tonnes',
        'Profit_INR',
        'Water_Efficiency_t_per_1000m3',
        'Disease_Pest_Risk_pct'
    ]
].mean()
print(overall_season)
```

```python
plt.figure(figsize=(12,6))
sns.heatmap(
    overall_season,
    annot=True,
    fmt='.2f'
)
plt.title('Overall Agricultural Performance Across Seasons')
plt.xlabel('Performance Indicators')
plt.ylabel('Season')
plt.show()
```
